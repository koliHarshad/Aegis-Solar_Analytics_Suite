```python
import pandas as pd
import numpy as np
```

```python
col_names = ["Year", "Day", "Hour", "B_Mag", "Bz", "Density", "Speed", "Kp_Index", "Sunspots"]
df = pd.read_csv('omni_daily_24years.csv', delim_whitespace=True, names=col_names, comment='#')
```

```python
df.head(10)
```

```python
df.info()
```

```python
df.describe()
```

```python
df = df.replace([999, 999.9, 99.9, 9999], np.nan)
```

```python
df.describe()
```

```python
df['Date'] = pd.to_datetime(df['Year'].astype(str) + df['Day'].astype(str), format='%Y%j')
```

```python
df['Kp_Index'] = df['Kp_Index'] / 10.0
```

```python
df.head(10)
```

```python
df[['Speed', 'Kp_Index', 'Sunspots', 'Bz']].corr()
```

```python
final_df = df[['Date', 'Speed', 'Density', 'Kp_Index', 'Sunspots', 'Bz']]
final_df.to_csv('solarstorm_clean.csv', index=False)
```
